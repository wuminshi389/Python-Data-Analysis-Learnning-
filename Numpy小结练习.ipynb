```python

'''
生成一个(5,5)的随机数组,范围[0,20)。
- 找出数组中大于 10 的元素。
- 将所有大于 10 的元素替换为 0。
'''
import numpy as np
arr = np.random.randint(0,20,(5,5))
print(arr)
print("大于10的元素为：",arr[arr>10])
arr = np.where(arr > 10, 0, arr)
print("将所有大于10的元素替换为0后为：", arr)
print(arr)
```
